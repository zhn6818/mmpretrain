```python
from mmpretrain import list_models, inference_model
```

```python
list_models(task='Image Classification', pattern='resnet18')
```

```python
# 待输入图像路径
img_path = 'demo/bird.JPEG'
```

```python
# 模型
model = 'resnet18_8xb32_in1k'
```

```python
# 推理预测
result = inference_model(model, img_path)
```
